## Key Decision

Lowered beta for improving functions (F5: β=1.5), kept high for flat ones (F1: β=3.0). Per-function tuning cut wasted queries compared to uniform settings.

## Lesson Learned

Early uniform β=2.5 for all functions failed. Manual tuning worked better than grid search at this data scale. Overfitting risk limited how aggressive I could tune.
